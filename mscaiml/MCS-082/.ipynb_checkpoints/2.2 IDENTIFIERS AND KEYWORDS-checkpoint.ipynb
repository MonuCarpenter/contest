Creating a function 

```python
def area(w, l): 
    return w * l

print(area(190.1001, 200.0))
```

Function with type 

```python
def area(w: int, l:int) -> int: 
    return w * l

print(area(19.01, 22.11))
```

```python
s1 = "hey" 
s2 = "varmil" 

s3 = s1 + " "+ s2 

if(s1 == "hey" and s2 == "varmil"):
    print(s3)

print("theek hain") if s1 == "hey" else print("not theek")

print(s3) 
```

```python
def calculate_area(width, height):
    """ Calculate the area of a rectangle.
    Arguments:
    width -- the width of the rectangle
    height -- the height of the rectangle
    """
    return width * height

print(calculate_area(20,3))
print(calculate_area.__doc__)
```
